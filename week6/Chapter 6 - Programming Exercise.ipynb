```python
# IMPORT STATEMENTS
# 1. Import math
import math
# FUNCTION DEFINITIONS
# 1. Define area(radius): returns pi * r squared
def area(radius):
  result = math.pi * radius ** 2
  return result
# 2. Define circumference(radius): returns 2 * pi * r
def circumference(radius):
  result = 2 * math.pi * radius
  return result
# 3. Define volume(radius, height):
def volume(radius, height):
#    - Call area(radius) and multiply by height.
  result = area(radius) * height
#    - Return the result.
  return result

# PROGRAM LOGIC
# 1. Input for radius (float)
radius = float(input("Enter radius: "))
# 2. Call and print area and circumference.
area_result = area(radius)
circumference_result = circumference(radius)
print("Area:", area_result)
print("circumference:", circumference_result)
# 3. Input for height (float)
height = float(input("Enter height: "))
# 4. Call and print volume.
volume_result = volume(radius, height)
print("volume:", volume_result)
```
