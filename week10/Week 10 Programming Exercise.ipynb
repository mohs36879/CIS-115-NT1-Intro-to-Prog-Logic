```python
# NAME: roster.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# add_players function
def add_players(roster):
  # - accept a list as a parameter
  # - start a while loop that keeps running until user types "quit"
  while True:
    # - ask the user to enter a player name
    name = input("Enter player name (or 'quit' to stop): ")
    # - if user enters "quit", stop the loop
    if name == "quit":
      break
    # - otherwise, add the name to the list using append
    roster.append(name)

# is_on_roster function
# - accept the list and a name to search
# - loop through the list
# - if name is found, return True
# - if loop ends without finding name, return False

# PROGRAM LOGIC
# - create empty list called roster
roster = []
# - call add_players to fill the list
add_players(roster)
# - ask user for one name to search for
# - call is_on_roster with roster and search name
# - if True, print "Player found!"
# - if False, print "Player not found."
# - print roster to test that names were added
print(roster)
```
