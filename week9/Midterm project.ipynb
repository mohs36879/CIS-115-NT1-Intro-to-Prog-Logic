```python
# rentalcar.py

# FUNCTION DEFINITIONS

# get_class_name(code)
def get_class_name(code):
  if code == "B":
    return "Budget"
  elif code == "S":
    return "Standard"
  else:
    return "Luxury"
# get_daily_rate(code)
def get_daily_rate(code):
  if code == "B":
    return 40
  elif code == "S":
    return 60
  else:
    return 80
# calculate_total_miles(start, end)
def calculate_total_miles(start, end):
  return end - start
# calculate_charged_miles(total_miles)
def calculate_charged_miles(total_miles):
  if total_miles > 100:
    return total_miles - 100
  else:
    return 0
# calculate_total_bill(code, days, charged_miles)
def calculate_total_bill(code, days, charged_miles):
  rate = get_daily_rate(code)
  base_cost = rate * days
  mileage_cost = charged_miles * 0.35
  total = base_cost + mileage_cost
  return total


# PROGRAM LOGIC

# get user input
code = input("Enter classification code (B, S, L): ")
code = code.upper()
# validate classification
while code != "B" and code != "S" and code != "L":
  print("Invalid code. Enter B, S, or L.")
  code = input("Enter classification code (B, S, L): ")
  code = code.upper()
# get odometer reading
start = int(input("Start odometer : "))
end = int(input("End odometer : "))
# get rental days
days = int(input("Days Rented: "))
# compute miles
total_miles = calculate_total_miles(start, end)
# compute charged miles
charged_miles = calculate_charged_miles(total_miles)
# compute bill
bill = calculate_total_bill(code, days, charged_miles)
# print summary
print("Total Miles:", total_miles)
print("Charged Miles:", charged_miles)
print("Total Bill:", bill)
```
