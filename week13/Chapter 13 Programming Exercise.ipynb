```python
# NAME: journal.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# write_entry function
def write_entry(filename, entry):
  # - accepts filename and entry
  # - open file in append mode ('a')
  file = open(filename, 'a')
  # - write entry with newline
  file.write(entry + '\n')
  # - close the file
  file.close()
  # confirm save
  print('Entry saved.')

# read_journal function
# - accepts filename
# - open file in read mode ('r')
# - read entire contents
# - print contents to screen
# - handle error if file does not exist

# PROGRAM LOGIC
# set filename (journal.txt)
filename = 'journal.txt'
# start while loop
while True:
  #   ask user: Write, Read, or Quit
  choice = input(" (W)rite, (R)ead, or (Q)uit? ")

  # if user chooses Write:
  if choice == 'W'or choice == 'w':
    # ask for entry
    entry = input("Enter your entry: ")
    # call write_entry
    write_entry(filename, entry)

  # if user chooses Read:
    # call read_journal

  # if user chooses Quit:
  elif choice == 'Q' or choice == 'q':
    # exit loop
    break

```
