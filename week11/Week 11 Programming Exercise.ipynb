```python
# NAME: gradebook.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
# Function: add_students
# Purpose: Add student names and grades to the dictionary
def add_students(gradebook):
# Steps:
  # - Start a loop that continues until the user types "done"
  while True:
    # - Ask the user to enter a student name
    name = input("Enter student name (or 'done' to finish): ")
    # - If the name is "done", exit the loop
    if name == "done":
      break
    # - Otherwise, ask for the student's grade (0-100)
    grade = int(input("Enter " + name + "'s grade: "))
    # - Add the name as a key and the grade as a value in the dictionary
    gradebook[name] = grade
# Function: calculate_average
# Purpose: Calculate the average of all grades in the dictionary
def calculate_average(gradebook):
# Steps:
    # - Check if the dictionary is empty
    if len(gradebook) == 0:
        return 0
    #   -If empty, return 0 to avoid division by zero
    # - Create a variable to store the total
    total = 0
    # - Loop through all values in the dictionary
    for grade in gradebook.values():
        # - Add each grade to the total
        total = total + grade
    # - Divide the total by the number of students
    average = total / len(gradebook)
    # - Return the average
    return average

# PROGRAM LOGIC

# Steps:
# - Create an empty dictionary called gradebook
gradebook = {}
# - Call the add_students function to fill the dictionary
add_students(gradebook)
# - Call the calculate_average function and store the result
average = calculate_average(gradebook)
# - Print a report header
print("\n--- Class Report ---")
# - Loop through the dictionary
for student in gradebook:
  #   - Print each student's name and grade
  print(student + ": " + str(gradebook[student]))
# - print the class average formatted to two decimal places
print("Class average: {:.2f}".format(average))
```
