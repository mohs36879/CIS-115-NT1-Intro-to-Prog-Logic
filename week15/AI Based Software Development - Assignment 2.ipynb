```python
import json
import os

DATA_FILE = "game_data.json"


# ----------------------------
# SAFE DATA HANDLING
# ----------------------------
def load_data():
    """Safely load JSON data or return empty dict if broken."""
    if not os.path.exists(DATA_FILE):
        return {}

    try:
        with open(DATA_FILE, "r") as file:
            data = json.load(file)

            # Ensure it's a dictionary
            if not isinstance(data, dict):
                return {}

            return data

    except (json.JSONDecodeError, IOError):
        print("[Warning] Data file corrupted. Starting fresh.")
        return {}


def save_data(data):
    """Safely save data to JSON."""
    try:
        with open(DATA_FILE, "w") as file:
            json.dump(data, file, indent=4)
        print("\n[System] Progress saved successfully.")
    except IOError:
        print("[Error] Could not save file.")


# ----------------------------
# INPUT SAFETY HELPERS
# ----------------------------
def get_positive_float(prompt):
    """Keeps asking until user enters valid positive number."""
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("❌ Value cannot be negative.")
                continue
            return value
        except ValueError:
            print("❌ Invalid input. Enter a number.")


def get_choice():
    """Ensures valid menu selection."""
    valid_choices = {"1", "2", "3", "4", "5"}
    choice = input("Select an option: ").strip()

    if choice in valid_choices:
        return choice

    print("❌ Invalid menu choice.")
    return None


# ----------------------------
# MAIN PROGRAM
# ----------------------------
def main():
    game_sessions = load_data()

    while True:
        print("\n--- Game Session Tracker ---")
        print("1. Log a Game Session")
        print("2. View Stats (Total, Average, Most Played)")
        print("3. Save and Exit")
        print("4. Edit or Delete a Game")
        print("5. Search for a Game")

        choice = get_choice()
        if not choice:
            continue

        # ----------------------------
        # 1. LOG GAME
        # ----------------------------
        if choice == "1":
            game_name = input("Enter game name: ").strip()

            if not game_name:
                print("❌ Game name cannot be empty.")
                continue

            hours = get_positive_float(f"Hours played {game_name}: ")

            game_sessions[game_name] = game_sessions.get(game_name, 0) + hours
            print(f"✔ Logged {hours} hours for {game_name}")

        # ----------------------------
        # 2. STATS
        # ----------------------------
        elif choice == "2":
            if not game_sessions:
                print("No data available yet.")
                continue

            total_hours = sum(game_sessions.values())
            avg_hours = total_hours / len(game_sessions) if game_sessions else 0

            most_played = max(game_sessions, key=game_sessions.get)

            print("\n--- STATS ---")
            for game, hours in game_sessions.items():
                print(f"- {game}: {hours} hours")

            print("\nSummary:")
            print(f"Total Games: {len(game_sessions)}")
            print(f"Total Hours: {total_hours}")
            print(f"Average Hours: {avg_hours:.2f}")
            print(f"Most Played: {most_played} ({game_sessions[most_played]} hrs)")

        # ----------------------------
        # 3. SAVE & EXIT
        # ----------------------------
        elif choice == "3":
            save_data(game_sessions)
            print("Goodbye!")
            break

        # ----------------------------
        # 4. EDIT / DELETE
        # ----------------------------
        elif choice == "4":
            game = input("Enter game name: ").strip()

            if game not in game_sessions:
                print("❌ Game not found.")
                continue

            action = input("Type 'edit' or 'delete': ").strip().lower()

            if action == "edit":
                new_hours = get_positive_float("Enter new total hours: ")
                game_sessions[game] = new_hours
                print("✔ Game updated.")

            elif action == "delete":
                del game_sessions[game]
                print("✔ Game deleted.")

            else:
                print("❌ Invalid action.")

        # ----------------------------
        # 5. SEARCH
        # ----------------------------
        elif choice == "5":
            search = input("Enter game name: ").strip()

            if search in game_sessions:
                print(f"{search}: {game_sessions[search]} hours")
            else:
                print("❌ Game not found.")


if __name__ == "__main__":
    main()
```
