```python
# Gaming Tracker Tool - v1.0

sessions = []

def add_session():
    game = input("Enter game name: ")

    try:
        hours = float(input("Enter hours played: "))
    except ValueError:
        print("Invalid input. Please enter a number.")
        return

    session = {"game": game, "hours": hours}
    sessions.append(session)
    print("Session added!\n")

def view_sessions():
    if not sessions:
        print("No sessions recorded.\n")
        return

    print("\n--- Game Sessions ---")
    for i, s in enumerate(sessions, start=1):
        print(f"{i}. {s['game']} - {s['hours']} hours")
    print()

def total_hours():
    total = sum(s["hours"] for s in sessions)
    print(f"\nTotal hours played: {total}\n")

def menu():
    while True:
        print("=== Gaming Tracker ===")
        print("1. Add Session")
        print("2. View Sessions")
        print("3. Total Hours")
        print("4. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
            add_session()
        elif choice == "2":
            view_sessions()
        elif choice == "3":
            total_hours()
        elif choice == "4":
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Try again.\n")

menu()
```
