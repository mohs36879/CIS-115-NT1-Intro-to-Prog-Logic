```python
import json
import os

# Constants
DATA_FILE = "game_data.json"

def load_data():
    """Load data from the JSON file. If file doesn't exist, return an empty dict."""
    if os.path.exists(DATA_FILE):
        with open(DATA_FILE, "r") as file:
            return json.load(file)
    return {}

def save_data(data):
    """Save the current session data to a JSON file."""
    with open(DATA_FILE, "w") as file:
        json.dump(data, file, indent=4)
    print("\n[System] Progress saved successfully.")

def main():
    # Load existing data
    game_sessions = load_data()

    while True:
        print("\n--- Game Session Tracker ---")
        print("1. Log a Game Session")
        print("2. View Stats (Total, Average, Most Played)")
        print("3. Save and Exit")
        print("4. Edit or Delete a Game")
        print("5. Search for a Game")

        choice = input("Select an option: ")

        # 1. Log Game Session
        if choice == "1":
            game_name = input("Enter the game name: ").strip()
            try:
                hours = float(input(f"How many hours did you play {game_name}? "))

                if game_name in game_sessions:
                    game_sessions[game_name] += hours
                else:
                    game_sessions[game_name] = hours

                print(f"Logged {hours} hours for {game_name}.")
            except ValueError:
                print("Error: Please enter a valid number for hours.")

        # 2. View Stats
        elif choice == "2":
            if not game_sessions:
                print("No data available yet.")
                continue

            print("\n--- Your Stats ---")
            most_played = None
            max_hours = 0

            total_hours = sum(game_sessions.values())
            average_hours = total_hours / len(game_sessions)

            for game, total in game_sessions.items():
                print(f"- {game}: {total} hours")
                if total > max_hours:
                    max_hours = total
                    most_played = game

            print(f"\nTotal Games Tracked: {len(game_sessions)}")
            print(f"Total Hours Played: {total_hours}")
            print(f"Average Hours per Game: {average_hours:.2f}")
            print(f"Most Played Game: {most_played} ({max_hours} hours)")

        # 3. Save and Exit
        elif choice == "3":
            save_data(game_sessions)
            print("Goodbye!")
            break

        # 4. Edit or Delete
        elif choice == "4":
            game_name = input("Enter the game to edit/delete: ").strip()

            if game_name in game_sessions:
                action = input("Type 'edit' or 'delete': ").lower()

                if action == "edit":
                    try:
                        new_hours = float(input("Enter new total hours: "))
                        game_sessions[game_name] = new_hours
                        print("Game updated successfully.")
                    except ValueError:
                        print("Invalid number.")

                elif action == "delete":
                    del game_sessions[game_name]
                    print("Game deleted successfully.")

                else:
                    print("Invalid action.")

            else:
                print("Game not found.")

        # 5. Search
        elif choice == "5":
            search = input("Enter game name to search: ").strip()

            if search in game_sessions:
                print(f"{search}: {game_sessions[search]} hours")
            else:
                print("Game not found.")

        else:
            print("Invalid choice, try again.")

if __name__ == "__main__":
    main()
```
