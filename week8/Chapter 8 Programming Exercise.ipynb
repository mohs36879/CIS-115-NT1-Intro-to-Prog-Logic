```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
#Define a function named is_palindrome that takes one parameter (word)
def is_palindrome(word):
  word = word.lower()
  reversed_word = word[::-1]
  if word == reversed_word:
    return True
  else:
    return False
#Convert the word to lowercase so the check is case insensitive
#Reverse the word using slicing
#Compare the lowercase word to the reversed word
#If they are the same, return True
#Otherwise, ruturn Flase

# PROGRAM LOGIC
#Ask the user to enter a word
user_word = input("Enter a word to check if it's a palindrome: ")
#Call the is_palindrome function and store the result
result = is_palindrome(user_word)
#If the result is True, print that it is a palindrome
if result:
  print(user_word + " is a palindrome!")
#If the result is False, print that it is not a palindrome
else:
  print(user_word + " is not a palindrome.")
```
